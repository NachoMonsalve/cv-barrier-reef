**YAML file**

```python
import yaml

def create_yolo_v10_yaml(yolo_folder: str, output_path: str, class_names: list):
    """
    Create a YOLOv10-compatible YAML file for training.

    Args:
        yolo_folder (str): Root folder containing YOLO data (e.g., images and labels).
        output_path (str): Path to save the YAML file.
        class_names (list): List of class names.
    """
    yaml_content = {
        "train": f"{yolo_folder}/images/train",  # Path to training images
        "val": f"{yolo_folder}/images/val",      # Path to validation images
        "nc": len(class_names),                  # Number of classes
        "names": {i: name for i, name in enumerate(class_names)}  # Map class IDs to names
    }

    # Write YAML file
    with open(output_path, "w") as yaml_file:
        yaml.dump(yaml_content, yaml_file, default_flow_style=False)
    print(f"YOLOv10 YAML file saved to {output_path}")
```

```python
# Define the dataset root folder and output YAML path
yolo_folder = "../data/yolo"  # Path to YOLO dataset folder
yaml_output_path = "../data/yolo/data.yaml"  # Path to save the YAML file
class_names = ["starfish"]  # Define your class names

# Generate YAML for YOLOv10
create_yolo_v10_yaml(yolo_folder, yaml_output_path, class_names)
```
